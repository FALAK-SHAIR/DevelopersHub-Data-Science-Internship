```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Iris dataset load karna (seaborn in alredy  built-in )
df = sns.load_dataset('iris')

# Dataset ka structure dekhna
print("Dataset size:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\n First 5 rows:")
df.head()
```

```python
# Cell 2 - Dataset ki detail summary
print("Dataset ki info:")
print(df.info())

print("\nStatistical summary:")
df.describe()
```

```python
# Cell 3 - Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='petal_length', hue='species', palette='Set1')
plt.title('Sepal Length vs Petal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```

```python
# Cell 4 - Histogram
df.hist(figsize=(10, 6), bins=20, color='steelblue', edgecolor='black')
plt.suptitle('Column Distribution', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Cell 5 - Box Plot
plt.figure(figsize=(12, 5))

for i, column in enumerate(['sepal_length', 'sepal_width', 'petal_length', 'petal_width']):
    plt.subplot(1, 4, i+1)
    sns.boxplot(data=df, x='species', y=column, palette='Set2')
    plt.title(column)
    plt.xticks(rotation=45)

plt.suptitle('Species measurements', y=1.02)
plt.tight_layout()
plt.show()
```
